````python
# Did initial setup and created cache folders (only needed once)

import os
from pathlib import Path

# Created local cache folders so files were not downloaded again and again
CACHE_ROOT = Path("/content/cache")
TORCH_HOME = CACHE_ROOT / "torch"
HF_HOME = CACHE_ROOT / "hf"
XDG_CACHE_HOME = CACHE_ROOT / "xdg"
TORCH_EXTENSIONS_DIR = CACHE_ROOT / "torch_extensions"

for p in [CACHE_ROOT, TORCH_HOME, HF_HOME, XDG_CACHE_HOME, TORCH_EXTENSIONS_DIR]:
    p.mkdir(parents=True, exist_ok=True)

# Set environment variables so PyTorch and HuggingFace used local cache
os.environ["TORCH_HOME"] = str(TORCH_HOME)
os.environ["HF_HOME"] = str(HF_HOME)
os.environ["TRANSFORMERS_CACHE"] = str(HF_HOME / "transformers")
os.environ["HF_HUB_CACHE"] = str(HF_HOME / "hub")
os.environ["XDG_CACHE_HOME"] = str(XDG_CACHE_HOME)
os.environ["TORCH_EXTENSIONS_DIR"] = str(TORCH_EXTENSIONS_DIR)

# Installed all required dependencies in one place
!pip -q install --upgrade onnx onnxscript onnxruntime-gpu ninja
!pip -q install --upgrade opencv-python
!pip -q install -U "transformers>=4.46" "accelerate" "bitsandbytes" "sentencepiece"

# Imported libraries and did a quick version check
import torch
import torchvision
import onnx
import onnxscript

print("torch:", torch.__version__)
print("torchvision:", torchvision.__version__)
print("onnx:", onnx.__version__)
print("onnxscript:", onnxscript.__version__)
print("cuda available:", torch.cuda.is_available())

# Exported DeepLabV3 model to ONNX

ONNX_PATH = Path("deeplabv3_logits.onnx")

if not ONNX_PATH.exists():

    # Created a wrapper model to return only logits
    class M(torch.nn.Module):
        def __init__(self):
            super().__init__()
            self.model = torchvision.models.segmentation.deeplabv3_resnet50(
                weights="DEFAULT"
            ).eval()

        def forward(self, x):
            return self.model(x)["out"]  # returned only logits

    model = M().eval()
    dummy = torch.randn(1, 3, 520, 520)

    # Exported the model to ONNX format
    torch.onnx.export(
        model,
        dummy,
        str(ONNX_PATH),
        opset_version=18,   # used opset 18 to avoid conversion issues
        input_names=["input"],
        output_names=["logits"],
    )

    print("Saved", ONNX_PATH)

else:
    # Skipped export if file already existed
    print("Found existing", ONNX_PATH, "- skipping export")

# Preloaded local LLM (DeepSeek-Coder)

USE_LOCAL_LLM = True

if USE_LOCAL_LLM:

    from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

    model_id = "deepseek-ai/deepseek-coder-6.7b-instruct"

    # Used 4-bit quantization to reduce memory usage
    bnb = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_use_double_quant=True,
        bnb_4bit_compute_dtype=torch.float16,
    )

    # Loaded tokenizer and model
    tokenizer = AutoTokenizer.from_pretrained(model_id, trust_remote_code=True)

    llm = AutoModelForCausalLM.from_pretrained(
        model_id,
        device_map="auto",
        quantization_config=bnb,
        trust_remote_code=True,
    )

    # Printed device info to confirm where model was loaded
    print(
        "LLM loaded on:",
        next(llm.parameters()).device,
        next(llm.parameters()).dtype
    )
````

````python
# Took image input using OpenCV and ran ONNX inference here

import cv2
import numpy as np
import onnxruntime as ort
from pathlib import Path
from google.colab.patches import cv2_imshow  # used this only for displaying in Colab

IMG_PATH = "img.jpeg"
ONNX_PATH = "deeplabv3_logits.onnx"

# Checked if required files were present
assert Path(IMG_PATH).exists(), f"Missing {IMG_PATH}"
assert Path(ONNX_PATH).exists(), f"Missing {ONNX_PATH} (run Step 0)"

# 1) Loaded image using OpenCV (it loads in BGR format by default)
bgr = cv2.imread(IMG_PATH, cv2.IMREAD_COLOR)
assert bgr is not None, "cv2.imread failed (check IMG_PATH)"

# Resized image to 520x520 to match model input size
bgr = cv2.resize(bgr, (520, 520), interpolation=cv2.INTER_LINEAR)

print("Input (OpenCV BGR):")
cv2_imshow(bgr)

# 2) Converted BGR to RGB because model expects RGB
rgb_u8 = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)

# 3) Preprocessed image (converted to float, normalized, changed layout)
x = rgb_u8.astype(np.float32) / 255.0  # scaled pixel values to [0,1]
x = x.transpose(2, 0, 1)               # converted HWC -> CHW

# Used ImageNet mean and std for normalization
mean = np.array([0.485, 0.456, 0.406], dtype=np.float32)[:, None, None]
std  = np.array([0.229, 0.224, 0.225], dtype=np.float32)[:, None, None]

x = ((x - mean) / std)[None, ...]      # added batch dimension (NCHW)

# 4) Ran ONNX inference (preferred CUDA if available)
sess = ort.InferenceSession(
    ONNX_PATH,
    providers=["CUDAExecutionProvider", "CPUExecutionProvider"]
)

# Got input tensor name from model
input_name = sess.get_inputs()[0].name

# Ran model to get logits
logits = sess.run(None, {input_name: x})[0]  # shape: (1,21,520,520)

print("logits:", logits.shape, logits.dtype)

# 5) Kept image name consistent for next CUDA step
image_u8 = rgb_u8  # (520,520,3) uint8 RGB
````

````python
# Generated CUDA kernel using LLM and ran it on GPU

import re
import numpy as np
import torch
from torch.utils.cpp_extension import load_inline

# Checked if previous steps were already run
assert "tokenizer" in globals() and "llm" in globals(), \
    "Step 0 was not run (tokenizer and llm missing)"

assert "logits" in globals() and "image_u8" in globals(), \
    "Step 1 was not run (logits and image missing)"

assert torch.cuda.is_available(), "CUDA was not available on this runtime"

# 1) Asked LLM to generate CUDA kernel with fixed constraints
prompt = r"""
Return ONLY one ```cpp``` code block.

Write CUDA kernel with exact signature:
__global__ void k(const float* logits, const unsigned char* img,
                  unsigned char* out, unsigned char* mask,
                  int C,int H,int W,int target);

One thread per pixel idx in [0,H*W).
Argmax over c using logits[c*(H*W) + idx].
mask[idx] = best (uint8).
If best==target copy 3 RGB bytes else zeros.
No helper functions.
"""

messages = [
    {"role": "system", "content": "You write correct, compilable CUDA kernels."},
    {"role": "user", "content": prompt},
]

# Formatted prompt for chat-style model
text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
) if hasattr(tokenizer, "apply_chat_template") else prompt

inputs = tokenizer([text], return_tensors="pt").to(llm.device)

# Generated kernel code from LLM
with torch.inference_mode():
    out_ids = llm.generate(**inputs, max_new_tokens=1000, do_sample=False)

resp = tokenizer.decode(
    out_ids[0, inputs["input_ids"].shape[1]:],
    skip_special_tokens=True
).strip()

print("=== LLM raw output ===")
print(resp)

# Extracted only cpp block from LLM output
m = re.search(r"```cpp\s*(.*?)```", resp, flags=re.DOTALL)
assert m, "LLM did not return a cpp block. Re-run this cell."

kernel_cpp = m.group(1).strip()

print("\n=== Extracted kernel code ===")
print(kernel_cpp)

# 2) Used same C++ binding as before
cpp_src = r"""
#include <torch/extension.h>
#include <vector>

std::vector<torch::Tensor> seg_apply_cuda(torch::Tensor logits,
                                         torch::Tensor img,
                                         int64_t target);

std::vector<torch::Tensor> seg_apply(torch::Tensor logits,
                                     torch::Tensor img,
                                     int64_t target) {
    return seg_apply_cuda(logits, img, target);
}

PYBIND11_MODULE(TORCH_EXTENSION_NAME, m) {
    m.def("seg_apply", &seg_apply, "seg_apply");
}
"""

# 3) Injected LLM-generated kernel into CUDA source
cuda_src = f"""
#include <torch/extension.h>
#include <cuda.h>
#include <cuda_runtime.h>
#include <vector>

{kernel_cpp}

std::vector<torch::Tensor> seg_apply_cuda(torch::Tensor logits,
                                         torch::Tensor img,
                                         int64_t target) {{
    int C = (int)logits.size(1);
    int H = (int)logits.size(2);
    int W = (int)logits.size(3);

    auto out = torch::empty(
        {{H, W, 3}},
        torch::TensorOptions().dtype(torch::kUInt8).device(logits.device())
    );

    auto mask = torch::empty(
        {{H, W}},
        torch::TensorOptions().dtype(torch::kUInt8).device(logits.device())
    );

    int HW = H * W;
    int threads = 256;
    int blocks = (HW + threads - 1) / threads;

    k<<<blocks, threads>>>(
        logits.data_ptr<float>(),
        img.data_ptr<unsigned char>(),
        out.data_ptr<unsigned char>(),
        mask.data_ptr<unsigned char>(),
        C, H, W, (int)target
    );

    return {{out, mask}};
}}
"""

# 4) Compiled and loaded CUDA extension
# If kernel code was changed, name was updated to force rebuild
ext = load_inline(
    name="seg_ext_min",
    cpp_sources=cpp_src,
    cuda_sources=cuda_src,
    functions=None,
    with_cuda=True,
    extra_cflags=["-O2"],
    extra_cuda_cflags=["-O2"],
    verbose=True,
)

print("Extension built successfully")

# 5) Ran kernel on current logits and image
logits_t = torch.from_numpy(logits).cuda().contiguous()     # (1,21,520,520)
image_t  = torch.from_numpy(image_u8).cuda().contiguous()  # (520,520,3)

TARGET = 15
out_t, mask_t = ext.seg_apply(logits_t, image_t, TARGET)

print("out_t:", tuple(out_t.shape), out_t.dtype, out_t.device)
print("mask_t:", tuple(mask_t.shape), mask_t.dtype, mask_t.device)
````

````python
# Displayed CUDA outputs and verified them using CPU reference

import numpy as np
import cv2
from google.colab.patches import cv2_imshow

# Checked if previous steps were already completed
assert "logits" in globals() and "image_u8" in globals()
assert "out_t" in globals() and "mask_t" in globals()
assert "TARGET" in globals()

# Moved CUDA outputs back to CPU as numpy arrays
out_np  = out_t.detach().cpu().numpy()     # (H,W,3) uint8 RGB masked image
mask_np = mask_t.detach().cpu().numpy()    # (H,W)   uint8 class ids

# Converted RGB to BGR because OpenCV expects BGR
out_bgr = cv2.cvtColor(out_np, cv2.COLOR_RGB2BGR)

print("Masked output (CUDA):")
cv2_imshow(out_bgr)

# Created a colored visualization of the mask (just for demo purpose)
mask_vis = cv2.applyColorMap(
    (mask_np * 10).astype(np.uint8),
    cv2.COLORMAP_JET
)

print("Mask visualization (CUDA):")
cv2_imshow(mask_vis)

# Ran a simple CPU reference using argmax
ref_mask = logits[0].argmax(axis=0).astype(np.uint8)

# Applied the same mask logic on CPU
ref_out = image_u8.copy()
ref_out[ref_mask != TARGET] = 0

# Compared CUDA output with CPU result
print("mask equal:", np.array_equal(ref_mask, mask_np))
print("out equal :", np.array_equal(ref_out, out_np))

# Saved outputs locally for submission
cv2.imwrite("masked_cuda.png", out_bgr)
cv2.imwrite("mask_ids.png", mask_np)
cv2.imwrite("mask_vis.png", mask_vis)

print("saved: masked_cuda.png, mask_ids.png, mask_vis.png")
````
